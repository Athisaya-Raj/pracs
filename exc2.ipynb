```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input
from tensorflow.keras.utils import load_img, img_to_array

img_path = r"D:\ImageNet\sample.jpg"
img = load_img(img_path, target_size=(224, 224))
x = preprocess_input(np.expand_dims(img_to_array(img), axis=0))

model = ResNet50(weights="imagenet")

layers = ["conv1_conv", "conv2_block3_out", "conv4_block6_out"]
feature_model = tf.keras.Model(model.input, [model.get_layer(l).output for l in layers])
features = feature_model.predict(x)

plt.imshow(img)
plt.title("Input Image")
plt.axis("off")
plt.show()

for name, feature in zip(layers, features):
    print(name, "Feature map dimensions:", feature.shape)

    plt.figure(figsize=(12, 6))
    for i in range(min(8, feature.shape[-1])):
        plt.subplot(2, 4, i + 1)
        plt.imshow(feature[0, :, :, i], cmap="viridis")
        plt.axis("off")
    plt.suptitle(name + " Feature Maps")
    plt.show()

    layer = model.get_layer(name)
    weights = layer.get_weights()

    if weights and len(weights[0].shape) == 4:
        filters = weights[0]
        print("Filter dimensions:", filters.shape)

        plt.figure(figsize=(12, 6))
        for i in range(min(8, filters.shape[-1])):
            f = filters[:, :, :, i]
            f = (f - f.min()) / (f.max() - f.min() + 1e-8)
            plt.subplot(2, 4, i + 1)
            plt.imshow(f)
            plt.axis("off")
        plt.suptitle(name + " Convolutional Filters")
        plt.show()
```
